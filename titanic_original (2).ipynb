CREATING A FUNCTION TO CLASSIFY AGE INTO AGE GROUPS

```python
def agegroups(x):
    if 0<=x<=10:
        return "0-10"
    elif 10<x<=20:
        return "11-20"
    elif 20<x<=30:
        return "21-30"
    elif 30<x<=40:
        return "31-40" 
    elif 40<x<=50:
        return "41-50"
    elif 50<x<=60:
        return "51-60"
    elif 60<x<=70:
        return "61-70"
    elif 70<x<=80:
        return "71-80"
    elif 80<x<=90:
        return "81-90"
    elif 90<x<=100:
        return "91-100"    
```

A NEW COLUMN CALLED "newgroup" HAS BEEN CREATED

```python
df['newgroup']=df['Age'].apply(agegroups)
df
```

CHECKING PRIORITY AGE GROUPS

```python
sns.countplot(data=df,x='newgroup',hue='Survived')
```

 CHILDREN AND ADULTS WERE GIVEN PRIORITY
